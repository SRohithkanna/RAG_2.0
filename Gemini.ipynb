```python
import os
from dotenv import load_dotenv

load_dotenv()  # loads variables from a local .env file if present

GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
if not GEMINI_API_KEY:
    raise ValueError(
        "GEMINI_API_KEY not found. Set it in a .env file or as an environment variable."
    )
```

```python
import google.generativeai as genai

genai.configure(api_key=GEMINI_API_KEY)

gemini_model = genai.GenerativeModel(
    model_name="models/gemini-3.6-flash",
    generation_config={"temperature": 0.1}
)

print("✅ Gemini Flash model ready")
```

```python
def call_gemini_flash(query, context):
    """
    Gemini call for RAG (NO WEB SEARCH)
    """

    context_text = "\n\n".join(
        f"[Source: {c['metadata']['pdf_name']} | Page {c['metadata']['page']}]\n"
        f"{c['text']}"
        for c in context
    )

    prompt = f"""
You are a medical question-answering assistant.

RULES:
- Use ONLY the information in the CONTEXT.
- Do NOT use external knowledge.
- If the answer is missing, say:
  "Information not found in the provided documents."
- Be concise and factual.

CONTEXT:
{context_text}

QUESTION:
{query}

ANSWER:
"""

    # ✅ NO tools → NO web search
    response = gemini_model.generate_content(prompt)
    return response.text.strip()
```
